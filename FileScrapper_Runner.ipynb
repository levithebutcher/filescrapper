# ⚡ File & Streaming Scraper Telegram Bot
### Google Colab 24/7 Cloud Runner

This notebook sets up, configures, and runs **FileScrapper Bot** in the cloud.

**Quick Guide:**
1. Fill in your `BOT_TOKEN` in **Step 1** (or add it to Colab Secrets using the 🔑 key icon).
2. Click **Runtime ➜ Run all** (or press `Ctrl+F9`).
3. Open Telegram, search for your bot, and send `/start`!

```python
# @title ⚙️ Step 1: Configuration & Credentials { display-mode: "form" }
import os

# --- Telegram Bot Credentials ---
# Option A: Enter token below in the form box
# Option B: Add BOT_TOKEN in Google Colab Secrets (Key icon on left sidebar)
BOT_TOKEN_FORM = ""  # @param {type:"string"}

# Try reading from Colab Secrets first
BOT_TOKEN = BOT_TOKEN_FORM
try:
    from google.colab import userdata
    secret_token = userdata.get("BOT_TOKEN")
    if secret_token:
        BOT_TOKEN = secret_token
except Exception:
    pass

BOT_TOKEN = BOT_TOKEN.strip()

# --- Repository & Scraper Settings (Pre-filled) ---
REPO_URL = "https://github.com/levithebutcher/filescrapper.git"  # @param {type:"string"}
BRANCH = "main"  # @param {type:"string"}
BASE_URL = "https://dropmms.com"  # @param {type:"string"}
MAX_POSTS_PER_FORUM = 25  # @param {type:"integer"}
POST_SEND_DELAY = 1.5  # @param {type:"number"}
REQUEST_TIMEOUT = 15  # @param {type:"integer"}
STREAMING_DOMAINS = "vidara.to,streamtape.com,streamfiles.net,lplaygo.com,doodstream.com,dood.,streamwish,filelions,dropload,mixdrop,streamruby,cloudfam.io"  # @param {type:"string"}

if not BOT_TOKEN:
    print("⚠️ ATTENTION: BOT_TOKEN is empty! Please enter your token in the form box or in Colab Secrets.")
else:
    print(f"✅ Bot Token loaded: {BOT_TOKEN[:8]}...{BOT_TOKEN[-4:]}")
print(f"📦 Repository: {REPO_URL} (branch: {BRANCH})")
print(f"🌐 Target Base URL: {BASE_URL}")
```

```python
# @title 🛡️ Step 2: Colab Safe Keep-Alive (No CPU freeze) { display-mode: "form" }
from IPython.display import display, Javascript

# Clean, lightweight keep-alive (Zero audio, 0% CPU usage, prevents browser tab freezing)
safe_js = """
if (window.colabKeepAliveTimer) {
    clearInterval(window.colabKeepAliveTimer);
}
window.colabKeepAliveTimer = setInterval(function() {
    var now = new Date().toLocaleTimeString();
    console.log("[Keep-Alive Heartbeat] Colab session active at " + now);
}, 60000);
"""
display(Javascript(safe_js))
print("✅ Safe Keep-Alive Activated! (Chrome will stay smooth & responsive)")
```

```python
# @title 📥 Step 3: Clone Code & Configure .env { display-mode: "form" }
import os

APP_DIR = "/content/filescrapper"

if os.path.exists(APP_DIR):
    print("🔄 Repository already cloned, pulling latest changes...")
    !cd {APP_DIR} && git fetch origin && git checkout {BRANCH} && git pull origin {BRANCH}
else:
    print(f"🚀 Cloning {REPO_URL}...")
    !git clone -b {BRANCH} {REPO_URL} {APP_DIR}

# Create .env file for the bot
env_lines = [
    f"BOT_TOKEN={BOT_TOKEN}",
    f"BASE_URL={BASE_URL}",
    f"MAX_POSTS_PER_FORUM={MAX_POSTS_PER_FORUM}",
    f"POST_SEND_DELAY={POST_SEND_DELAY}",
    f"REQUEST_TIMEOUT={REQUEST_TIMEOUT}",
    f"STREAMING_DOMAINS={STREAMING_DOMAINS}",
]
with open(f"{APP_DIR}/.env", "w", encoding="utf-8") as f:
    f.write("\n".join(env_lines) + "\n")

print(f"✅ Configuration (.env) created successfully at {APP_DIR}/.env")
```

```python
# @title 📦 Step 4: Install Dependencies { display-mode: "form" }
!cd /content/filescrapper && pip install -q --upgrade-strategy only-if-needed -r requirements.txt curl_cffi cloudscraper
print("✅ Dependencies successfully installed!")
```
